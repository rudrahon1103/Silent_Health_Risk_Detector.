 # Stroke Dataset
## Exploratory Data Analysis (EDA)

```python
stroke = pd.read_csv("../datasets/healthcare-dataset-stroke-data.csv")
```

```python
stroke.head()
```

```python
stroke.tail()
```

```python
stroke.shape
```

```python
stroke.columns
```

```python
stroke.info()
```

```python
stroke.describe()
```

```python
stroke.isnull().sum()
```

```python
stroke.duplicated().sum()
```

```python
stroke.columns = stroke.columns.str.strip()
stroke = stroke.drop_duplicates()
```

```python
stroke['bmi'] = stroke['bmi'].fillna(stroke['bmi'].median())
```

```python
plt.figure(figsize=(8,5))
plt.hist(stroke['age'], bins=20, edgecolor='black')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(stroke['bmi'], bins=20, edgecolor='black')
plt.title("BMI Distribution")
plt.xlabel("BMI")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='stroke', data=stroke)
plt.title("Stroke Cases")
plt.xlabel("Stroke")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.scatter(stroke['age'], stroke['bmi'])
plt.title("Age vs BMI")
plt.xlabel("Age")
plt.ylabel("BMI")
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(stroke.corr(numeric_only=True), cmap='coolwarm', annot=False)
plt.title("Correlation Heatmap")
plt.show()
```

```python
stroke['stroke'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(6,6)
)
plt.title("Stroke Distribution")
plt.ylabel("")
plt.show()
```
